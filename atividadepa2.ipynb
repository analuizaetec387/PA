```python
numero1 = int (input("digite um numero: "))
numero2 = int (input("digite um numero: "))

soma = numero1 + numero2
subtração = numero1 - numero2
divisão = numero1 / numero2
multiplicação = numero1 * numero2

print (f"o resultado da soma de {numero1} e {numero2} é igual: {soma}")
print (f"o resultado da subtração de {numero1} e {numero2} é igual: {subtração}")
print (f"o resultado da divisão de {numero1} e {numero2} é igual: {divisão}")
print (f"o resultado da multiplicação de {numero1} e {numero2} é igual: {multiplicação}")
```

```python
# 4

nota1 = float (input("Digite a nota do primeiro bimestre: "))

if nota1 < 0 or nota1 > 10:
  print ("nota inválida, digite uma nota de 0 a 10:")
  nota1 = float (input("Digite a nota do primeiro bimestre: "))

nota2 = float (input("Digite a nota do segundo bimestre: "))

if nota2 < 0 or nota1 > 10:
  print ("nota inválida, digite uma nota de 0 a 10:")
  nota2 = float (input("Digite a nota do segundo bimestre: "))

nota3 = float (input("Digite a nota do terceiro bimestre: "))

if nota3 < 0 or nota1 > 10:
  print ("nota inválida, digite uma nota de 0 a 10:")
  nota2 = float (input("Digite a nota do terceiro bimestre: "))

nota4 = float (input("Digite a nota do quarto bimestre: "))

if nota4 < 0 or nota1 > 10:
  print ("nota inválida, digite uma nota de 0 a 10:")
  nota4 = float (input("Digite a nota do quarto bimestre: "))


media = (nota1 + nota2 + nota3 + nota4)/4

if media < 5:
 print (f"reprovado, sua média é {media}")

elif media <= 7:
  print (f"recuperação, sua média é {media}")

else:
  print (f"aprovado, sua média é {media}")
```
