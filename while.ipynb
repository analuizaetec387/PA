```python
# 7

soma = 0
numero = int(input("Digite um numero: "))

while numero != 0:
  soma = soma + numero
  numero = int(input("Digite um numero: "))

print("Soma:", soma)
```

```python
# 8

maior = 0

numero = int(input("Digite um número: "))

while numero != 0:
    if numero > maior:
        maior = numero

    numero = int(input("Digite um número: "))

print("Maior número:", maior)
```

```python
# 9

soma = 0
contador = 1

while contador <=5:
  nota = float(input("Digite a nota: "))
  soma = soma + nota
  contador = contador + 1

media = soma / 5

print("Media:", media)


```

```python
# 10

saldo = 1000
opcao = 0

while opcao != 4:
    print("Escolha uma opção:")
    print("1 - Saldo")
    print("2 - Depósito")
    print("3 - Saque")
    print("4 - Sair")

    opcao = int(input("Digite a opção desejada: "))
    if opcao == 1:
        print(f"Saldo: R${saldo}")
    elif opcao == 2:
        deposito = float(input("Digite o valor do depósito: "))
        saldo += deposito
    elif opcao == 3:
        saque = float(input("Digite o valor do saque: "))
        if saque > saldo:
            print("Saldo insuficiente!")
    elif opcao == 4:
        print("Saindo")
    else:
        print("Opção inválida!")
```
