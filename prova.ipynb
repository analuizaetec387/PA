```python
# 4

numero = int(input("Digite um número inteiro positivo: "))

if numero < 0:
    print("Número negativo digitado. Fatorial não existe para números negativos.")
elif numero == 0:
    print("O número digitado é 0. O fatorial de 0 é 1.")
else:
    print(f"O número digitado é positivo: {numero}")
    factorial = 1
    calculation_str = str(numero)
    for i in range(numero - 1, 0, -1):
        factorial *= (i + 1)
        calculation_str += f" x {i}"


    if numero == 1:
        factorial = 1
        calculation_str = "1"

    print(f"{numero}! = {calculation_str} = {factorial}")
```
