```python
# 4

dia = input("digite um dia da semana: ")

match dia:
  case "segunda-feira":
    print("dia de semana")

  case "terça-feira":
    print("dia de semana")

  case "quarta-feira":
    print("dia de semana")

  case "quinta-feira":
    print("dia de semana")

  case "sexta-feira":
    print("dia de semana")

  case "sabado | domingo":
    print("final de semana")
```

```python
nota = input("Digite a nota (A, B ou C:")

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```
