```python
# 6

matriz = [
    [3, 6, 9],
    [12, 15, 18],
    [21, 24, 27]
]

coluna = []
for linha in matriz:
  coluna.append(linha[2])

print(coluna)
```

```python
# 7

matriz = [
    [1, 2],
    [3, 4],
    [5, 6]
]

for linha in matriz:
    for numero in linha:
        print(numero)
```

```python
# 8

matriz = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

soma = 0

for linha in matriz:
    soma += linha[1]

print("Soma:", soma)
```

```python
# 9

matriz = [
    [12, 5, 8],
    [20, 3, 15],
    [7, 25, 10]
]

maior = matriz[0][0]

for linha in matriz:
    for numero in linha:
        if numero > maior:
            maior = numero

print("Maior número:", maior)
```
